## 构建 Reinforce 网络架构
本实现基于以下两个经典版本实现：
- [PyTorch 官方强化学习示例](https://github.com/pytorch/examples/blob/main/reinforcement_learning/reinforce.py)
- [Udacity Reinforce 实现](https://github.com/udacity/deep-reinforcement-learning/blob/master/reinforce/REINFORCE.ipynb)
- [由 Chris1nexus 改进的融合版本](https://github.com/huggingface/deep-rl-class/pull/95)

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit6/reinforce.png" alt="Reinforce"/>

我们期望：
- 两个全连接层（fc1 和 fc2）。
- 使用 ReLU 作为 fc1 的激活函数。
- 使用 Softmax 输出动作上的概率分布。

````python
class Policy(nn.Module):
    def __init__(self, s_size, a_size, h_size):
        super(Policy, self).__init__()
        # 创建两个全连接层
        self.fc1 = nn.Linear(in_features=s_size, out_features=h_size)
        self.fc2 = nn.Linear(in_features=h_size, out_features=a_size)



    def forward(self, x):
        # 定义前向传播
        # 状态输入 fc1 然后应用 ReLU 激活函数
        out1 = nn.ReLU(self.fc1(x))
        # fc1 的输出输入 fc2
        out2 = self.fc2(out1)
        # 最后输出 softmax
        return nn.Softmax(out2, dim=1)

    def act(self, state):
        """
        给定一个状态，采取动作
        """
        state = torch.from_numpy(state).float().unsqueeze(0).to(device)
        probs = self.forward(state).cpu()
        m = Categorical(probs)
        action = np.argmax(m)
        return action.item(), m.log_prob(action)
````

### （真正的）解决方案

````python
class Policy(nn.Module):
    def __init__(self, s_size, a_size, h_size):
        super(Policy, self).__init__()
        self.fc1 = nn.Linear(s_size, h_size)
        self.fc2 = nn.Linear(h_size, a_size)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return F.softmax(x, dim=1)

    def act(self, state):
        state = torch.from_numpy(state).float().unsqueeze(0).to(device)
        probs = self.forward(state).cpu() # shape(1,2)
        m = Categorical(probs)   # 一个包含 1 个独立样本批次（Batch）、每个批次有 2 个可能类别的分类分布对象
        action = m.sample()  # 正确姿势：使用 sample 方法从概率分布中采样一个动作
        return action.item(), m.log_prob(action)
````

通过使用 CartPole，调试变得更加简单，因为**我们确信 bug 来自于我们的代码编写，而不是由于复杂的环境导致的**。

- 既然**我们希望从动作概率分布中进行采样**，我们就不能使用 `action = np.argmax(m)`，因为它总是只输出概率最高的那个动作。

- 我们需要替换为 `action = m.sample()`，它会根据概率分布 P(.|s) 随机采样一个动作。

### 构建 Reinforce 训练算法
这是 Reinforce 算法的伪代码：

<img src="https://huggingface.co/datasets/huggingface-deep-rl-course/course-images/resolve/main/en/unit6/pg_pseudocode.png" alt="Policy gradient pseudocode"/>

- 当我们计算回报 $G_t$（第 6 行）时，我们发现我们是在计算**从时间步 t 开始**的折扣奖励总和。

- 为什么？因为我们的策略应该**只依据后续的后果来强化动作**：所以在采取某个动作之前获得的奖励是无用的（因为它们不是由当前动作引起的），**只有动作发生之后的奖励才起作用**。

- 在编写代码之前，你应该阅读 [don't let the past distract you](https://spinningup.openai.com/en/latest/spinningup/rl_intro3.html#don-t-let-the-past-distract-you) 这一章节，它解释了为什么我们要使用只针对未来的回报（reward-to-go）策略梯度。

我们使用由 [Chris1nexus](https://github.com/Chris1nexus) 编写的非常巧妙的技巧来**高效地计算每个时间步的回报**。注释解释了具体步骤，你也可以[查阅 PR 详细说明](https://github.com/huggingface/deep-rl-class/pull/95)。
但总的思路是**高效地在每个时间步计算回报**。

你可能会问的第二个问题是：**为什么我们要最小化损失函数（minimize the loss）**？你之前不是说要用梯度上升（Gradient Ascent）而不是梯度下降吗？

- 我们的目标是最大化效用函数 $J(\theta)$，但在 PyTorch 和 Tensorflow 等框架中，更习惯于**最小化一个目标函数（Loss）**。
    - 假设我们想在某个时间步强化动作 3。训练前该动作的概率 P 是 0.25。
    - 我们希望修改 $\theta$ 使得 $\pi_\theta(a_3|s; \theta) > 0.25$。
    - 因为所有概率之和必须为 1，最大化 $\pi_\theta(a_3|s; \theta)$ 就会自动**最小化其他动作的概率**。
    - 我们可以让 PyTorch 去**最小化 $1 - \pi_\theta(a_3|s; \theta)$**。
    - 当 $\pi_\theta(a_3|s; \theta)$ 接近 1 时，此损失函数接近 0。
    - 因此，我们通过这个最小化损失的过程来实现最大化 $\pi_\theta(a_3|s; \theta)$ 的梯度优化。

````python
def reinforce(policy: Policy, optimizer, n_training_episodes, max_t, gamma, print_every):
    # 用于在训练过程中计算滑动平均分数
    scores_deque = deque(maxlen=100)
    scores = []
    # 对应伪代码第 3 行
    for i_episode in range(1, n_training_episodes+1):
        saved_log_probs = []
        rewards = []  # rewards[t] = 第 t 个时间步的即时奖励
        state, _ = env.reset() # TODO: 重置环境
        # 对应伪代码第 4 行（在环境中执行一回合）
        for t in range(max_t):
            action, log_prob = policy.act(state) # TODO: 获取动作
            saved_log_probs.append(log_prob)
            state, reward, terminated, truncated, _ = env.step(action) # TODO: 在环境中执行一步
            rewards.append(reward)
            if terminated or truncated:
                break
        scores_deque.append(sum(rewards))
        scores.append(sum(rewards))

        # 对应伪代码第 6 行：计算回报
        returns = deque(maxlen=max_t)
        n_steps = len(rewards)
        # 计算每个时间步的折扣回报，
        # 即：时间步 t 的 gamma 折扣回报 (G_t) + 时间步 t 的即时奖励之和

        # 时间复杂度为 O(N)，其中 N 是总时间步数
        # （这个折扣回报 G_t 的定义遵循 Sutton&Barto 2017 第二版草案第 44 页的定义）
        # G_t = r_(t+1) + r_(t+2) + ...

        # 在此定义下，每个时间步 t 的回报可以重用已计算的未来回报 G_(t+1) 来进行计算：
        # G_t = r_(t+1) + gamma*G_(t+1)
        # G_(t-1) = r_t + gamma* G_t
        # （这遵循动态规划方法，通过记忆化中间结果来避免重复计算）

        # 这样做是正确的，因为这等价于（参见 Sutton&Barto 2017 第二版草案第 46 页）：
        # G_(t-1) = r_t + gamma*r_(t+1) + gamma*gamma*r_(t+2) + ...


        ## 综上，我们计算时间步 t 的回报公式为：
        #               gamma[t] * return[t] + reward[t]
        #
        ## 我们从最后一个时间步倒序计算到第一个时间步，以便应用上述递推公式
        ## 并避免从前到后计算时所产生的冗余计算。

        ## 因此，“returns” 双端队列将按时间顺序保存从 t=0 到 t=n_steps 的回报，
        ## 借助 appendleft() 函数，我们可以以 O(1) 的常数时间将其插入到队首，
        ## 而普通的 Python 列表则需要 O(N) 的时间。
        for t in range(n_steps)[::-1]:
            disc_return_t = (returns[0] if len(returns)>0 else 0)
            returns.appendleft(gamma * disc_return_t + rewards[t]) # TODO: 在这里完成代码

        ## 对回报进行归一化（标准化）处理，以使训练更加稳定
        eps = np.finfo(np.float32).eps.item()

        ## eps 是单精度浮点数能表示的最小值，
        # 将其加到回报的标准差中以避免发生除零等数值不稳定问题
        returns = torch.tensor(returns)
        returns = (returns - returns.mean()) / (returns.std() + eps)

        # 对应第 7 行：
        policy_loss = []
        for log_prob, disc_return in zip(saved_log_probs, returns):
            policy_loss.append(-log_prob * disc_return)
        policy_loss = torch.cat(policy_loss).sum()

        # 对应第 8 行：PyTorch 默认进行梯度下降，所以加负号
        optimizer.zero_grad()
        policy_loss.backward()
        optimizer.step()

        if i_episode % print_every == 0:
            print('Episode {}\tAverage Score: {:.2f}'.format(i_episode, np.mean(scores_deque)))

    return scores
````

#### 答案参考

````python
def reinforce(policy, optimizer, n_training_episodes, max_t, gamma, print_every):
    # 用于在训练过程中计算滑动平均分数
    scores_deque = deque(maxlen=100)
    scores = []
    # 对应伪代码第 3 行
    for i_episode in range(1, n_training_episodes+1):
        saved_log_probs = []
        rewards = []  # rewards[t] = 第 t 个时间步的即时奖励
        state, _ = env.reset()
        # 对应伪代码第 4 行
        for t in range(max_t):
            action, log_prob = policy.act(state)
            saved_log_probs.append(log_prob)
            state, reward, terminated, truncated, _ = env.step(action)
            rewards.append(reward)
            if terminated or truncated:
                break
        scores_deque.append(sum(rewards))
        scores.append(sum(rewards))

        # 对应伪代码第 6 行：计算回报
        returns = deque(maxlen=max_t)  # returns[t] 第 t 个时间步的后续剩余汇报
        n_steps = len(rewards)
        # 计算每个时间步的折扣回报，
        # 即：
        #      时间步 t 的 gamma 折扣回报 (G_t) + 时间步 t 的即时奖励之和
        #
        # 时间复杂度为 O(N)，其中 N 是总时间步数
        # （这个折扣回报 G_t 的定义遵循 Sutton&Barto 2017 第二版草案第 44 页的定义）
        # G_t = r_(t+1) + r_(t+2) + ...

        # 在此定义下，每个时间步 t 的回报可以重用已计算的未来回报 G_(t+1) 来进行计算：
        # G_t = r_(t+1) + gamma*G_(t+1)
        # G_(t-1) = r_t + gamma* G_t
        # （这遵循动态规划方法，通过记忆化中间结果来避免重复计算）

        # 这样做是正确的，因为这等价于（参见 Sutton&Barto 2017 第二版草案第 46 页）：
        # G_(t-1) = r_t + gamma*r_(t+1) + gamma*gamma*r_(t+2) + ...


        ## 综上，我们计算时间步 t 的回报公式为：
        #               gamma[t] * return[t] + reward[t]
        #
        ## 我们从最后一个时间步倒序计算到第一个时间步，以便应用上述递推公式
        ## 并避免从前到后计算时所产生的冗余计算。

        ## 因此，“returns” 双端队列将按时间顺序保存从 t=0 到 t=n_steps 的回报，
        ## 借助 appendleft() 函数，我们可以以 O(1) 的常数时间将其插入到队首，
        ## 而普通的 Python 列表则需要 O(N) 的时间。
        for t in range(n_steps)[::-1]:
            disc_return_t = (returns[0] if len(returns)>0 else 0)  # 第 t + 1 个时间步的后续剩余汇报
            returns.appendleft(gamma * disc_return_t + rewards[t])

        ## 对回报进行归一化（标准化）处理，以使训练更加稳定
        eps = np.finfo(np.float32).eps.item()
        ## eps 是单精度浮点数能表示的最小值，
        # 将其加到回报的标准差中以避免发生除零等数值不稳定问题
        returns = torch.tensor(returns)
        returns = (returns - returns.mean()) / (returns.std() + eps)

        # 对应第 7 行：
        policy_loss = []
        for log_prob, disc_return in zip(saved_log_probs, returns):
            policy_loss.append(-log_prob * disc_return)
        policy_loss = torch.cat(policy_loss).sum()

        # 对应第 8 行：PyTorch 默认进行梯度下降，所以加负号
        optimizer.zero_grad()
        policy_loss.backward()
        optimizer.step()

        if i_episode % print_every == 0:
            print('Episode {}\tAverage Score: {:.2f}'.format(i_episode, np.mean(scores_deque)))

    return scores
````

## 定义评估方法 📝
- 这里我们定义了用于测试 Reinforce 智能体性能的评估方法。

````python
def evaluate_agent(env, max_steps, n_eval_episodes, policy):
  """
  对智能体进行 ``n_eval_episodes`` 轮评估，并返回平均奖励和奖励标准差。
  :param env: 评估所用的环境
  :param n_eval_episodes: 评估智能体的局数
  :param policy: 待评估的 Reinforce 智能体
  """
  episode_rewards = []
  for episode in range(n_eval_episodes):
    state, _ = env.reset()
    step = 0
    done = False
    total_rewards_ep = 0

    for step in range(max_steps):
      action, _ = policy.act(state)
      new_state, reward, terminated, truncated, info = env.step(action)
      total_rewards_ep += reward

      if terminated or truncated:
        break
      state = new_state
    episode_rewards.append(total_rewards_ep)
  mean_reward = np.mean(episode_rewards)
  std_reward = np.std(episode_rewards)

  return mean_reward, std_reward
````

````python
def record_video(env, policy, out_directory, fps=30):
  """
  生成智能体游玩的录像视频
  :param env: 环境（render_mode 必须是 "rgb_array"）
  :param policy: 我们的智能体策略模型
  :param out_directory: 视频输出路径
  :param fps: 每秒帧数（对于 taxi-v3 和 frozenlake-v1 我们使用 1）
  """
  images = []
  done = False
  state, _ = env.reset()
  img = env.render()
  images.append(img)
  while not done:
    # 采取当前状态下输出概率最大的动作
    action, _ = policy.act(state)
    state, reward, terminated, truncated, info = env.step(action) # 更新状态以便进行下一步记录
    done = terminated or truncated
    img = env.render()
    images.append(img)
  imageio.mimsave(out_directory, [np.array(img) for i, img in enumerate(images)], fps=fps)
````

### 创建 Pixelcopter 环境

Colab 版通过 `gym-games` 获得 `Pixelcopter-PLE-v0`，但 gym-games 只支持老 gym。这里直接基于 PLE 写一个 gymnasium 版的等价环境：观测（PLE 的 7 维游戏状态）、动作集、奖励（PLE 默认奖励）和画面都与 gym-games 的实现一致，并以同一个 id 注册，下面 `gym.make(env_id)` 的写法保持不变。

导入 PLE 时会打印 `Couldn't import doom`，这是 PLE 在尝试加载可选的 Doom 游戏，可以忽略。

````python
from gymnasium import spaces
from ple import PLE
from ple.games.pixelcopter import Pixelcopter


class PixelcopterEnv(gym.Env):
    """gym-games 中 PixelcopterEnv 的 gymnasium 版，只负责把 PLE 的接口转成 gymnasium 接口。"""
    metadata = {"render_modes": ["rgb_array"], "render_fps": 30}

    def __init__(self, render_mode=None):
        self.render_mode = render_mode
        self.game = PLE(Pixelcopter(), fps=30, display_screen=False,
                        state_preprocessor=lambda s: np.array(list(s.values()), dtype=np.float32))
        self.game.init()
        self.action_set = self.game.getActionSet()
        self.action_space = spaces.Discrete(len(self.action_set))
        self.observation_space = spaces.Box(-np.inf, np.inf, shape=self.game.getGameStateDims(), dtype=np.float32)

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.game.reset_game()
        return self.game.getGameState(), {}

    def step(self, action):
        reward = self.game.act(self.action_set[action])
        # Pixelcopter 没有步数上限，truncated 恒为 False，与 gym-games 一致
        return self.game.getGameState(), reward, self.game.game_over(), False, {}

    def render(self):
        # getScreenRGB 是 (宽, 高, 3)，转成图像库约定的 (高, 宽, 3)
        return np.fliplr(np.rot90(self.game.getScreenRGB(), 3))


gym.register(id="Pixelcopter-PLE-v0", entry_point=PixelcopterEnv)
````

````python
env_id = "Pixelcopter-PLE-v0"
env = gym.make(env_id, render_mode="rgb_array")
eval_env = gym.make(env_id)
s_size = env.observation_space.shape[0]
a_size = env.action_space.n
````

### 定义新的策略网络 🧠
- 既然环境变得更加复杂，我们需要一个更深层的神经网络

````python
class Policy(nn.Module):
    def __init__(self, s_size, a_size, h_size):
        super(Policy, self).__init__()
        # 在这里定义三个网络层
        self.fc1 = nn.Linear(s_size, h_size)
        self.fc2 = nn.Linear(h_size, h_size*2)
        self.fc3 = nn.Linear(h_size*2, a_size)

    def forward(self, x):
        # 在这里定义前向传播计算过程
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return F.softmax(x, dim=1)

    def act(self, state):
        state = torch.from_numpy(state).float().unsqueeze(0).to(device)
        probs = self.forward(state).cpu()
        m = Categorical(probs)
        action = m.sample()
        return action.item(), m.log_prob(action)
````
